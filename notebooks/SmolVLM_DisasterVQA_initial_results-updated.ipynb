```python
import re

binary_results = results_df[
    results_df["question_type"] == "Binary"
].copy()

def normalize_yes_no(text):
    text = str(text).lower().strip()

    # Look for an explicit Yes/No answer
    matches = re.findall(r'\b(yes|no)\b', text)

    if matches:
        return matches[-1]

    return None

binary_results["predicted_answer"] = (
    binary_results["prediction"]
    .apply(normalize_yes_no)
)

binary_results["ground_truth_clean"] = (
    binary_results["ground_truth"]
    .astype(str)
    .str.lower()
    .str.strip()
)

binary_results["correct"] = (
    binary_results["predicted_answer"]
    == binary_results["ground_truth_clean"]
)

print("Binary questions:", len(binary_results))
print("Correct:", binary_results["correct"].sum())
print("Accuracy:",
      round(binary_results["correct"].mean() * 100, 2), "%")
```
